```python
from tensorflow.keras.models import load_model 
import tensorflow as tf
import sklearn
import pandas as pd 
import numpy as np 
import pickle
```

```python
model = load_model('model.h5') #the trained model is loaded 

# loading the encoder and the scaler

with open('one_hot_encoder_geo.pkl','rb') as file:
    label_encoder_geo = pickle.load(file)

with open('label_encoder_gender.pkl','rb') as file:
    label_encoder_gender = pickle.load(file)
    
with open('scaler.pkl','rb') as file:
    scaler = pickle.load(file)
    
    
```

```python
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    "Tenure": 3,
    'Balance': 60000,
    'NumOfProducts': 2, 
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 500000
}
```

```python
label_encoder_geo.get_feature_names_out
```

```python
geo_encoded = label_encoder_geo.transform([[input_data['Geography']]]).toarray()
geo_encoded_df  = pd.DataFrame(geo_encoded,columns=label_encoder_geo.get_feature_names_out())
geo_encoded_df
```

```python
input_df = pd.DataFrame([input_data])
```

```python
# combinig one-hot encoded columns with input data 
input_df = pd.concat([input_df.reset_index(drop=True),geo_encoded_df],axis=1)
input_df = input_df.drop(['Geography'],axis= 1)
```

```python
input_df
```

```python
# encoding the categorial variable 
data_gender = label_encoder_gender.transform([input_data['Gender']])
```

```python
input_df
```

```python
input_df['Gender']= label_encoder_gender.transform(input_df["Gender"])
```

```python
input_df
```

```python
# scaling the input data 
input_scaled = scaler.transform(input_df)
input_scaled
```

```python
prediction = model.predict(input_scaled)
prediction
```

```python
prediction_prob = prediction[0][0]
```

```python
prediction_prob
```

```python
if prediction_prob > 0.5:
    print("The customer is likely to churn.")
else:
    print("The customer is not likely to churn")
```
